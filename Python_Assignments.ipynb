## Assignment 5.2
**Question:** WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.

```python
# Assignment 5.2: Sum numbers divisible by 7 and 9 from 1 to n
n = int(input("Enter a number: "))

total = 0
for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        total += i
        print(f"{i} is divisible by both 7 and 9")

print(f"\nSum of numbers divisible by 7 and 9 from 1 to {n} = {total}")
```

## Assignment 5.3
**Question:** WAP to add all prime numbers from 1 to n and n is given by the user.

```python
# Assignment 5.3: Sum all prime numbers from 1 to n
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num // 2 + 1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter a number: "))

total = 0
primes = []
for i in range(1, n + 1):
    if is_prime(i):
        total += i
        primes.append(i)

print(f"Prime numbers from 1 to {n}: {primes}")
print(f"Sum of prime numbers from 1 to {n} = {total}")
```

## Assignment 6.1
**Question:** WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
# Assignment 6.1: Function to sum all odd numbers from 1 to n
def sum_odd_numbers(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total

n = int(input("Enter a number: "))
result = sum_odd_numbers(n)
print(f"Sum of odd numbers from 1 to {n} = {result}")
```

## Assignment 6.2
**Question:** WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
# Assignment 6.2: Function to sum all prime numbers from 1 to n
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num // 2 + 1):
        if num % i == 0:
            return False
    return True

def sum_prime_numbers(n):
    total = 0
    for i in range(1, n + 1):
        if is_prime(i):
            total += i
    return total

n = int(input("Enter a number: "))
result = sum_prime_numbers(n)
print(f"Sum of prime numbers from 1 to {n} = {result}")
```
