### Question 1: Lists

```python
roll_no = "1024170402"
L = [int(digit) * 10 for digit in roll_no]

print(L)

L.append(99)
print(L)
# append() added the value 99 to the very end of the list L, increasing its length by 1.

L.insert(2, 55)
print(L)
# insert() placed the value 55 exactly at index 2, shifting all subsequent elements one position to the right.

L.remove(55)
print(L)

L.pop()
print(L)

L.sort()
print(L)

L.sort(reverse=True)
print(L)

print(L[:3])
print(L[-3:])

avg = sum(L) / len(L)
filtered_L = [x for x in L if x > avg]
print(filtered_L)
```

### Question 2: Tuples

```python
scores = tuple([int(digit) * 10 for digit in "1024170402"][:8])

high_score = max(scores)
print("Highest:", high_score, "Index:", scores.index(high_score))
low_score = min(scores)
print("Lowest:", low_score, "Count:", scores.count(low_score))

reversed_scores = list(reversed(scores))
print(reversed_scores)
# Tuples cannot be reversed in place because tuple immutability prevents their elements from being changed or reordered after creation.

user_score = int(input("Enter a score: "))
if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Not present")

try:
    scores[0] = 100
except TypeError as e:
    print("Error:", type(e).__name__, "-", e)
# This happens because tuples are immutable and do not support item assignment, unlike lists which are mutable.

first, second, *remaining = scores
print("First:", first, "Second:", second, "Remaining:", remaining)
```

### Question 3: Random Numbers & Control Flow

```python
import random

random.seed(1024170402)

rand_list = [random.randint(100, 900) for _ in range(100)]

odds = [x for x in rand_list if x % 2 != 0]
print("Count of odd numbers:", len(odds))
print("Odd numbers:", odds)

evens = [x for x in rand_list if x % 2 == 0]
print("Count of even numbers:", len(evens))
print("Even numbers:", evens)

def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

primes = [x for x in rand_list if is_prime(x)]
print("Count of prime numbers:", len(primes))
print("Prime numbers:", primes)

counts = {}
for num in rand_list:
    counts[num] = counts.get(num, 0) + 1

most_freq_num = max(counts, key=counts.get)
print("Most frequent number:", most_freq_num, "Count:", counts[most_freq_num])
```

### Question 4: Sets

```python
roll_digits = [int(digit) for digit in "1024170402"[:8]]

A = {d * 7 for d in roll_digits}
B = {d * 9 for d in roll_digits}
print("Set A:", A)
print("Set B:", B)

print("Union:", A.union(B))

print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() returns elements present in one specific set but not the other, whereas symmetric_difference() returns elements in either set but not both.

print("Symmetric Difference:", A.symmetric_difference(B))

print("Is A subset of B:", A.issubset(B))
print("Is B superset of A:", B.issuperset(A))

val = int(input("Enter a value X to remove: "))
A.discard(val)
print("Set A after discard:", A)
# discard() is safer because it silently does nothing if the element is missing, whereas remove() raises a KeyError and crashes the program.
```

### Question 5: Dictionaries

```python
my_dict = {
    "name": "Ehsen Fazili",
    "roll_no": "1024170402",
    "branch": "Computer Science",
    "age": 22,
    "city": "Patiala"
}

my_dict["location"] = my_dict.pop("city")

my_dict["cgpa"] = 9.0

my_dict["age"] += 1

copy1 = my_dict.copy()
copy2 = my_dict.copy()
popped_val = copy1.pop("branch")
del copy2["branch"]
# pop() removes the specified key-value pair and returns the value, whereas the del keyword simply deletes the pair and returns nothing.

for k, v in my_dict.items():
    print(f"{k} -> {v}")

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key is not present. Safe fallback active.")

friend_dict = {
    "name": "Friend Name",
    "roll_no": "1024170403",
    "branch": "Electrical",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print(merged_dict)
# When both dictionaries share a key during unpacking, the value from the rightmost dictionary (dict2) "wins" and overwrites the preceding value.

str_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(str_dict)
```
