### a) Reverse the NumPy array

### b) Find the most frequent value and its indices
**i.** `x = np.array([1,2,3,4,5,1,2,1,1,1])`

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent_x = values[np.argmax(counts)]
indices_x = np.where(x == most_frequent_x)[0]

print("Most frequent value:", most_frequent_x)
print("Indices:", indices_x)
```

**ii.** `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values_y, counts_y = np.unique(y, return_counts=True)
most_frequent_y = values_y[np.argmax(counts_y)]
indices_y = np.where(y == most_frequent_y)[0]

print("Most frequent value:", most_frequent_y)
print("Indices:", indices_y)
```

## Q.4 1-D NumPy array with 25 evenly spaced numbers from 10 to 100
The array is named after the student (`ehsen`) as requested.

```python
ehsen = np.linspace(10, 100, 25)
ehsen
```

```python
print("Dimensions (ndim):", ehsen.ndim)
print("Shape:", ehsen.shape)
print("Total elements (size):", ehsen.size)
print("Data type:", ehsen.dtype)
print("Total bytes consumed (nbytes):", ehsen.nbytes)
```

### Transpose using `reshape()`
A 1-D array has only one axis, so there is nothing to swap — reshaping it to a column vector `(25, 1)` is the practical equivalent of "transposing" it into a 2-D column form.

```python
ehsen_transposed_reshape = ehsen.reshape(25, 1)
ehsen_transposed_reshape
```

### Can we do the same with the `.T` attribute?
Directly calling `.T` on the original 1-D array does **not** change anything — NumPy treats a 1-D array as its own transpose, so its shape stays `(25,)`. `.T` only swaps axes when there is more than one axis. If we first reshape it into 2-D (e.g. `(1, 25)`), then `.T` behaves like a true transpose and gives the same `(25, 1)` result as above.

```python
print("ehsen.T shape (no change on 1-D array):", ehsen.T.shape)

ehsen_2d = ehsen.reshape(1, 25)
ehsen_2d_T = ehsen_2d.T
print("Reshaped to (1,25) then .T shape:", ehsen_2d_T.shape)

print("Same result as reshape(25,1)?", np.array_equal(ehsen_2d_T, ehsen_transposed_reshape))
```

## Q.5 2-D Array of three rows and four columns (`ucs420_ehsen`)

```python
ucs420_ehsen = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])
ucs420_ehsen
```

### Compute mean, median, max, min, unique elements

```python
print("Mean:", ucs420_ehsen.mean())
print("Median:", np.median(ucs420_ehsen))
print("Max:", ucs420_ehsen.max())
print("Min:", ucs420_ehsen.min())
print("Unique elements:", np.unique(ucs420_ehsen))
```

### Reshape the array to four rows and three columns (`reshaped_ucs420_ehsen`)

```python
reshaped_ucs420_ehsen = ucs420_ehsen.reshape(4, 3)
reshaped_ucs420_ehsen
```

### Resize the array to two rows and three columns (`resized_ucs420_ehsen`)
`np.resize` fills the new shape from the flattened original array (repeating/truncating as needed) instead of requiring the same total number of elements as `reshape` does.

```python
resized_ucs420_ehsen = np.resize(ucs420_ehsen, (2, 3))
resized_ucs420_ehsen
```
