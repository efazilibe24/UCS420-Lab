```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
temperature
```

### a) Add 2°C to every reading using vectorization

### c) Identify all readings greater than 32°C using Boolean indexing

```python
above_32 = temperature[temperature > 32]
above_32
```

### d) Count how many readings exceed 32°C

```python
count_above_32 = np.sum(temperature > 32)
count_above_32
```

### e) Why vectorization and Boolean indexing are more efficient than an explicit loop
NumPy arrays store elements in contiguous blocks of memory as a fixed data type. Vectorized operations (like `temperature + 2`) are implemented in optimized, pre-compiled C code that operates on the whole array at once — there's no per-element Python bytecode interpretation, no repeated type-checking, and the CPU can use techniques like SIMD (processing several values per instruction) and better cache usage. A Python `for` loop, by contrast, re-enters the interpreter for every single element, which carries a lot of overhead.\n\nBoolean indexing works the same way: `temperature > 32` computes a single Boolean mask over the entire array in one vectorized pass, and then that mask is used to select matching elements in another vectorized pass — again avoiding an explicit Python-level loop with `if` checks. For large datasets this difference in speed becomes dramatic, and the code is also shorter and easier to read.

## Q.3 Slicing, views vs. copies, reshape, flatten/ravel

### a) Create the original array

```python
original = np.array([1, 2, 3, 4, 5, 6])
original
```

### b) Create a slice (index 1 to 4) named `subset`

```python
subset = original[1:5]
subset
```

### c) Modify the first element of `subset` to 999 and display both

```python
subset[0] = 999
print("original:", original)
print("subset:", subset)
```

A basic slice (`original[1:5]`) is a **view**, not a copy — it shares the same underlying memory buffer as `original`. Changing `subset[0]` therefore also changes `original[1]`, which is visible above.

### d) Create another slice using `.copy()`, modify it, and display both

```python
original_copy_slice = original[1:5].copy()
original_copy_slice[0] = 500

print("original:", original)
print("copied slice:", original_copy_slice)
```

Because `.copy()` allocates a brand-new array with its own memory, modifying `original_copy_slice` has **no effect** on `original` this time.

```python
raveled[0] = -1
print("raveled:", raveled)
print("original matrix:\n", matrix)
```

`ravel()` returns a **view** whenever possible (the matrix here is contiguous), so modifying `raveled[0]` also changes `matrix[0, 0]` — the original array **is** affected.

`flatten()` always returns a **copy**, so modifying `flattened[1]` leaves `matrix` completely unchanged.

## Q.4 Predicting a user's assistance score (Ordinary Least Squares)

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

### a) Shape and dimensions of X

```python
print("Shape of X:", X.shape)
print("Dimensions (ndim) of X:", X.ndim)
```

### b) X.T and what the transpose represents

```python
X_T = X.T
X_T
```

`X` has one row per user and one column per feature (sleep hours, activity level, stress level). `X.T` flips this layout so that it has one row per **feature** and one column per **user** — i.e. rows and columns are swapped. It's the same data, just reorganized so that each row now represents "all users' values for a single feature", which is exactly the shape needed for the matrix operations in the OLS formula below.

### c) Matrix product X.T @ X

```python
XtX = X.T @ X
XtX
```

### d) Matrix inverse using np.linalg.inv()

```python
XtX_inv = np.linalg.inv(XtX)
XtX_inv
```

### e) Implement the Ordinary Least Squares equation $\beta = (X^TX)^{-1}X^Ty$

```python
beta = XtX_inv @ X.T @ y
beta
```

### f) What do the resulting coefficients represent?

Each value in `beta` is the estimated weight for one of the three features (sleep hours, activity level, stress level) in a linear model:\n\n$$\text{assistance score} = \beta_1 \cdot \text{sleep} + \beta_2 \cdot \text{activity} + \beta_3 \cdot \text{stress}$$\n\nConcretely, `beta[0]` is how much the predicted assistance score changes for each extra hour of sleep (holding the other two features fixed), `beta[1]` is the change per unit increase in activity level, and `beta[2]` is the change per unit increase in stress level — with the sign indicating whether that feature pushes the score up or down.

### g) Predicted assistance score for a new user

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
predicted_score
```
