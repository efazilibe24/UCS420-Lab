## Q.4 Read a CSV file and display its first five rows.
Dataset: [Iris dataset](https://www.kaggle.com/datasets/uciml/iris).

Place `Iris.csv` in the same folder as this notebook (downloaded from the Kaggle link above) and run the cell below. If the file isn't found, the cell automatically generates an equivalent `Iris.csv` locally so the notebook still runs end-to-end.

```python
import os

if not os.path.exists("Iris.csv"):
    # Fallback: build an equivalent Iris.csv locally (same schema as the Kaggle version)
    from sklearn.datasets import load_iris
    iris = load_iris(as_frame=True)
    iris_df = iris.frame.copy()
    iris_df.columns = ["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm","target"]
    species_map = {0: "Iris-setosa", 1: "Iris-versicolor", 2: "Iris-virginica"}
    iris_df["Species"] = iris_df["target"].map(species_map)
    iris_df = iris_df.drop(columns=["target"])
    iris_df.insert(0, "Id", range(1, len(iris_df) + 1))
    iris_df.to_csv("Iris.csv", index=False)

iris_data = pd.read_csv("Iris.csv")
iris_data.head()
```

## Q.5 Delete row 4 and delete column 3 from the CSV loaded in Q.4. Display the result.

```python
iris_modified = iris_data.drop(index=4)          # delete row with label 4
iris_modified = iris_modified.drop(iris_modified.columns[3], axis=1)  # delete column at position 3
iris_modified.head()
```

## Q.6 Create a sample dataset (`employees.csv`)

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(emp_data)
employees["Joining_Date"] = pd.to_datetime(employees["Joining_Date"])
employees.to_csv("employees.csv", index=False)
employees
```

### e) Statistics

```python
avg_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("i.   Average salary:", avg_salary)
print("ii.  Total bonus paid:", total_bonus)
print("iii. Youngest employee's age:", youngest_age)
print("iv.  Highest performance rating:", highest_rating)
```

### f) Sort the DataFrame by Salary column in descending order

```python
employees_sorted = employees.sort_values(by="Salary", ascending=False)
employees_sorted
```

### g) Add a new column that categorizes employees based on performance rating

```python
def rating_category(r):
    if r >= 4.5:
        return "Excellent"
    elif r >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(rating_category)
employees
```
